```python
import numpy as np
from scipy.optimize import minimize


y = np.array([67, 32]) 
f = 500 
d = 1
phi = np.radians(15) 

def predict(P):
    Y = P[0]
    Z = P[1]
    y1_hat = f * Y / Z
    y2_hat = f * ((Y - d)*np.cos(phi) + (Z)*np.sin(phi)) / ((d-Y)*np.sin(phi)+Z*np.cos(phi))
    return np.array([y1_hat, y2_hat])

def objective_function(P):
    y_hat = predict(P)
    return np.sum((y - y_hat) ** 2)

initial_guess = [0.5, 3.2] # starting point for P
result = minimize(objective_function, initial_guess)

print("Optimized P:", result.x)
```
